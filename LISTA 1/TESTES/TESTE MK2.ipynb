```python
import numpy as np
from io import BytesIO
from urllib.request import urlopen
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

image_url = ("https://raw.githubusercontent.com/florencio300/""LISTA-1-PROC-SIN-II/main/dataset/lena3.tif")

with urlopen(image_url) as response:
    original = Image.open(BytesIO(response.read()))
    
rgb = np.asarray(original)

fig, axes = plt.subplots(1, 3, figsize=(12, 4), constrained_layout=True)

for ax, channel, name in zip(axes, range(3), ("Red", "Green", "Blue")):
    ax.imshow(rgb[:, :, channel], cmap="gray", vmin=0, vmax=255)
    ax.set_title(f"{name} channel")
    ax.axis("off")

for ax in axes:
    ax.axis("off")
    ax.set_box_aspect(1)

plt.show() 
```

```python

from io import BytesIO
from urllib.request import urlopen
import matplotlib.pyplot as plt
from PIL import Image

image = ("https://raw.githubusercontent.com/florencio300/""LISTA-1-PROC-SIN-II/main/dataset/lena3.tif")

with urlopen(image) as response:
    image = Image.open(BytesIO(response.read()))

image_R = image.getchannel("R")
image_G = image.getchannel("G")
image_B = image.getchannel("B")

fig, axes = plt.subplots(1, 3, figsize=(12, 4), constrained_layout=True)


axes[0].imshow(image_R, cmap="gray",)
axes[0].set_title("RED channel")

axes[1].imshow(image_G, cmap="gray")
axes[1].set_title("GREEN channel")

axes[2].imshow(image_B, cmap="gray")
axes[2].set_title("BLUE channel")

for ax in axes:
    ax.axis("off")
    ax.set_box_aspect(1)

plt.show() 
```
